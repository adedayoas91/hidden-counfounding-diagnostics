# Fast c-GC* extension metrics

Runs the shared simulation grid with the optimized c-GC* implementation and preserves each trial's data, ground truth, and per-depth adjacency matrices.

This is not a bootstrap notebook: `N_REPEATS` controls independent simulation trials, not bootstrap replicates. Progress is shown with an overall trial bar and a conditioning-depth bar for the active trial, followed by completion statistics, elapsed time, and saved paths.

```python
from pathlib import Path

root = next(
    path
    for path in (Path.cwd().resolve(), *Path.cwd().resolve().parents)
    if (path / 'src' / 'markovianity_diagnostic').exists()
)

METHOD = 'fast_gcstar_fcgc'
SCENARIOS = [
    'order1_unconfounded',
    'order3_unconfounded',
    'latent_common_driver',
    'variable_lag_unconfounded',
]
P_VALUES = [1, 2, 3, 4, 5, 6, 7]
N_REPEATS = 10
T = 2000
D = 10
SEED = 42
OUTPUT_DIR = root / 'outputs' / 'simulations' / 'extension_metrics' / METHOD

total_trials = len(SCENARIOS) * N_REPEATS
total_depth_fits = total_trials * len(P_VALUES)
print('=' * 72)
print('Configured c-GC* extension-metrics run')
print(f'Method registry key: {METHOD}')
print(f'Scenarios ({len(SCENARIOS)}): {SCENARIOS}')
print(f'Repeats per scenario: {N_REPEATS}')
print(f'Depth grid: {P_VALUES}')
print(f'Data dimensions: T={T}, d={D}; base seed={SEED}')
print(f'Total trials: {total_trials}')
print(f'Total single-depth fits: {total_depth_fits}')
print(f'Output directory: {OUTPUT_DIR}')
print('=' * 72)
```

```python
import subprocess
import sys
import time

command = [
    sys.executable,
    '-u',
    str(root / 'notebooks' / 'simulations' / 'run_extension_metrics.py'),
    '--method',
    METHOD,
    '--scenarios',
    *SCENARIOS,
    '--p-values',
    *[str(depth) for depth in P_VALUES],
    '--n-repeats',
    str(N_REPEATS),
    '--T',
    str(T),
    '--d',
    str(D),
    '--seed',
    str(SEED),
]

print('Starting extension metrics...')
print('Command:', ' '.join(command))
started = time.perf_counter()
subprocess.run(command, cwd=root, check=True)
elapsed = time.perf_counter() - started
print(f'✓ c-GC* extension metrics completed in {elapsed:.1f}s')
print(f'✓ Outputs: {OUTPUT_DIR}')
```
